```python
from pathlib import Path
import csv

student_id = "2025-0722"

folder_path = Path.home() / "Documents" / "Activity_5_Files"

txt_files = list(folder_path.glob("*.txt"))

summary = []

for file in txt_files:
    lines = file.read_text().splitlines()

    non_empty_lines = [line for line in lines if line.strip() != ""]

    line_count = len(non_empty_lines)
    word_count = sum(len(line.split()) for line in non_empty_lines)
    char_count = sum(len(line.replace(" ", "")) for line in non_empty_lines)

    words_per_line = word_count / line_count if line_count > 0 else 0
    chars_per_word = char_count / word_count if word_count > 0 else 0

    summary.append({
        "filename": file.name,
        "lines": line_count,
        "words": word_count,
        "chars": char_count,
        "wpl": words_per_line,
        "cpw": chars_per_word
    })

summary.sort(key=lambda x: x["wpl"], reverse=True)

print(f"{'Filename':<25} {'Lines':<6} {'Words':<6} {'Chars':<8} {'W/Line':<10} {'C/Word'}")
print("-" * 70)

for item in summary:
    print(f"{item['filename']:<25} {item['lines']:<6} {item['words']:<6} {item['chars']:<8} {item['wpl']:<10.2f} {item['cpw']:.2f}")

if summary:
    print("\nMost content-dense file:", summary[0]["filename"])
    print("Least content-dense file:", summary[-1]["filename"])

csv_path = folder_path / f"content_summary_{student_id}.csv"

with csv_path.open("w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Filename", "Lines", "Words", "Characters", "Words/Line", "Chars/Word"])

    for item in summary:
        writer.writerow([
            item["filename"],
            item["lines"],
            item["words"],
            item["chars"],
            f"{item['wpl']:.2f}",
            f"{item['cpw']:.2f}"
        ])

print(f"\nSummary exported to: {csv_path}")
```

```python
from pathlib import Path
import string
import csv
from collections import Counter

student_id = "2025-0722"

folder_path = Path.home() / "Documents" / "Activity_5_Files"

stopwords_path = folder_path / "stopwords.txt"

if stopwords_path.exists():
    stopwords = set(stopwords_path.read_text().lower().split())
else:
    stopwords = {"the", "is", "and", "to", "a", "of", "in", "on", "for", "with"}

txt_files = [f for f in folder_path.glob("*.txt") if f.name != "stopwords.txt"]

overall_counter = Counter()
file_counters = {}

for file in txt_files:
    text = file.read_text().lower()

    text = text.translate(str.maketrans("", "", string.punctuation))

    words = text.split()

    filtered_words = [word for word in words if word not in stopwords]

    counter = Counter(filtered_words)
    file_counters[file.name] = counter

    overall_counter.update(counter)

print("\nTop Meaningful Words (Overall):")
for word, count in overall_counter.most_common(10):
    print(f"{word}: {count}")

print("\nWord Frequencies Per File:")
for filename, counter in file_counters.items():
    print(f"\n{filename}:")
    for word, count in counter.most_common(5):
        print(f"  {word}: {count}")

csv_path = folder_path / f"word_frequency_{student_id}.csv"

with csv_path.open("w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Word", "Total Count"])

    for word, count in overall_counter.items():
        writer.writerow([word, count])

print(f"\nWord frequency data exported to: {csv_path}")
```

```python
from pathlib import Path
import shutil
from datetime import datetime

student_id = "2025-0722"
student_name = "Gavin Idrish Bolanos"

base_folder = Path.home() / "Documents" / "Activity_5_Exercises"
backup_folder = base_folder / f"backup_{student_id}"
backup_folder.mkdir(parents=True, exist_ok=True)

original_file = base_folder / f"intro_{student_id}.txt"

if not original_file.exists():
    original_file.write_text(f"Sample content for {student_name}")
    print("Sample file created.")

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

backup_filename = f"{original_file.stem}_{student_id}_{timestamp}.txt"
backup_path = backup_folder / backup_filename

shutil.copy2(original_file, backup_path)

file_size = original_file.stat().st_size

log_file = backup_folder / f"backup_log_{student_id}.txt"

with log_file.open("a") as log:
    log.write(f"Student ID: {student_id}\n")
    log.write(f"Student Name: {student_name}\n")
    log.write(f"Original File: {original_file.name}\n")
    log.write(f"Timestamp: {timestamp}\n")
    log.write(f"File Size: {file_size} bytes\n")
    log.write(f"Backup Path: {backup_path}\n")
    log.write("-" * 50 + "\n")

print(f"\nBackup completed for {student_id} ({student_name})")
print(f"File saved as: {backup_filename}")
print(f"Log updated at: {backup_folder}")
```
